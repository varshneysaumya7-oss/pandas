```python
import pandas as pd
import numpy as np

employee_df = pd.DataFrame({
    'Employee': ['Aman', 'Riya', 'Rahul', 'Neha', 'Karan'],
    'Department': ['HR', 'IT', 'Sales', 'Finance', 'IT'],
    'Experience': [2, 4, 3, 5, 1]
})

new_df = employee_df[['Employee', 'Experience']]

print(new_df)
```

```python
print(employee_df['Department'])

print(employee_df.columns)
```

```python
employee_df['Office'] = 'Head Office'

print(employee_df)
```

```python
employee_df['Office'] = np.arange(5)

print(employee_df)
```

```python
office_names = pd.Series(
    ['Delhi', 'Mumbai', 'Pune'],
    index=[0, 2, 4]
)

employee_df['Office'] = office_names

print(employee_df)
```

```python
del employee_df['Office']

print(employee_df)
```
